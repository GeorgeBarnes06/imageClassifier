```python
import numpy as np
import tensorflow as tf
from tensorflow.keras.models import load_model
import matplotlib.pyplot as plt
import os
```

```python
model = load_model('models/imageClassifier1.h5')
```

```python
from sklearn.metrics import accuracy_score, precision_score, recall_score, f1_score, roc_auc_score, confusion_matrix
```

```python
IMG_SIZE = (256, 256)
BATCH_SIZE = 32
test_path = os.path.join("data", "test")

test = tf.keras.preprocessing.image_dataset_from_directory(
    test_path,
    image_size=IMG_SIZE,
    batch_size=BATCH_SIZE,
    label_mode='binary',
    shuffle=False
)
```

```python
from sklearn.metrics import accuracy_score, precision_score, recall_score, f1_score, roc_auc_score, confusion_matrix

def evaluate_cnn_model(model, dataset, name="CNN Model"):
    """
    Evaluate a Keras model on image data
    
    Args:
        model: Trained Keras model
        dataset: tf.data.Dataset (like your test dataset)
        name: Model name for display
    """

    y_true = []
    y_pred_prob = []
    
    for images, labels in dataset:
        predictions = model.predict(images, verbose=0)
        y_pred_prob.extend(predictions.flatten()) 
        y_true.extend(labels.numpy())
    
    y_true = np.array(y_true)
    y_pred_prob = np.array(y_pred_prob)
    
    y_pred = (y_pred_prob > 0.5).astype(int)
    
    acc = accuracy_score(y_true, y_pred)
    prec = precision_score(y_true, y_pred)
    rec = recall_score(y_true, y_pred)
    f1 = f1_score(y_true, y_pred)
    auc = roc_auc_score(y_true, y_pred_prob)
    
    print(f"\n{name} Results:")
    print(f"Accuracy   : {acc:.4f}")
    print(f"Precision  : {prec:.4f}")
    print(f"Recall     : {rec:.4f}")
    print(f"F1-score   : {f1:.4f}")
    print(f"AUROC      : {auc:.4f}")
    return y_true, y_pred, y_pred_prob

# Use it like this:
y_test_true, y_test_pred, y_test_prob = evaluate_cnn_model(model, test, "Pneumonia CNN")
```
